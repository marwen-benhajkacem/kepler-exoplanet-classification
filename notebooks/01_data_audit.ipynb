# 01 - Data audit: Kepler Objects of Interest (KOI)
Structure, missing values, duplicates, target distribution and data-leakage risk. Source: NASA Exoplanet Archive (cumulative KOI table).

```python

# ==============================================
# ETAPE 1 : IMPORTATION DES BIBLIOTHEQUES
# ==============================================

import pandas as pd
import numpy as np

# Permet d'afficher toutes les colonnes dans Colab
pd.set_option("display.max_columns", None)

# Permet d'afficher les nombres avec une precision lisible
pd.set_option("display.float_format", "{:.3f}".format)

print("Bibliotheques chargees avec succes.")
```

```python
# Load the dataset from the repository (works locally, on Colab and on GitHub)
from pathlib import Path
DATA_PATH = next(p for p in [Path("../data/exoplanets.csv"), Path("data/exoplanets.csv")] if p.exists())
df_raw = pd.read_csv(DATA_PATH, low_memory=False)
print("Rows:", df_raw.shape[0], "| Columns:", df_raw.shape[1])
df = df_raw.copy()  # working copy, the original is never modified
```

```python

# ==============================================
# INSPECTION GENERALE DU DATASET
# ==============================================

# Afficher les cinq premieres observations
print("===== APERCU DES DONNEES =====")
display(df.head())

# Afficher les dimensions
print("\n===== DIMENSIONS =====")
print(df.shape)

# Afficher les noms de toutes les colonnes
print("\n===== NOMS DES COLONNES =====")
for i, colonne in enumerate(df.columns):
    print(i + 1, ":", colonne)

# Examiner les types et les valeurs non nulles
print("\n===== INFORMATIONS GENERALES =====")
df.info()

```

```python

# ==============================================
# ANALYSE DES VALEURS MANQUANTES
# ==============================================

# Nombre de valeurs manquantes par colonne
missing_count = df.isnull().sum()

# Pourcentage de valeurs manquantes
missing_percent = (missing_count / len(df)) * 100

# Construire un tableau recapitulatif
missing_report = pd.DataFrame({
    "Valeurs_manquantes": missing_count,
    "Pourcentage": missing_percent,
    "Type": df.dtypes.astype(str)
})

# Trier les colonnes de la plus incomplete
# a la moins incomplete
missing_report = missing_report.sort_values(
    by="Pourcentage",
    ascending=False
)

# Afficher uniquement les colonnes ayant des manquants
missing_report = missing_report[
    missing_report["Valeurs_manquantes"] > 0
]

print("===== RAPPORT DES VALEURS MANQUANTES =====")
display(missing_report)
```

```python

# ==============================================
# DETECTION DES DOUBLONS
# ==============================================

# Compter les lignes totalement identiques
duplicate_count = df.duplicated().sum()

print("Nombre de lignes dupliquees :", duplicate_count)

# Afficher quelques doublons si disponibles
if duplicate_count > 0:

    print("\nExemples de doublons :")

    display(
        df[df.duplicated(keep=False)].head(10)
    )
else:
    print("Aucun doublon exact detecte.")
```

```python

# ==============================================
# ANALYSE DE LA VARIABLE CIBLE
# ==============================================

target = "koi_disposition"

# Verifier que la cible existe
if target in df.columns:

    print("===== REPARTITION DES CLASSES =====")

    # Nombre d'observations par classe
    print(df[target].value_counts(dropna=False))

    # Pourcentage de chaque classe
    print("\n===== REPARTITION EN POURCENTAGE =====")

    print(
        df[target].value_counts(
            normalize=True,
            dropna=False
        ) * 100
    )

else:

    print("ATTENTION : koi_disposition est absente !")
    print("Nous devons verifier la version du fichier.")
```

```python
# Load the dataset from the repository (works locally, on Colab and on GitHub)
from pathlib import Path
DATA_PATH = next(p for p in [Path("../data/exoplanets.csv"), Path("data/exoplanets.csv")] if p.exists())
import pandas as pd, numpy as np, matplotlib.pyplot as plt
df_original = pd.read_csv(DATA_PATH)
df = df_original.copy(deep=True)  # never modify df_original
```

```python

# ============================================================
# CLASSIFICATION DES VARIABLES
# ============================================================

# 1. Identifiants : ne représentent pas des mesures physiques
identifiants = [
    "kepid",
    "kepoi_name",
    "kepler_name"
]

# 2. Variable cible
cible = "koi_disposition"

# 3. Variables présentant un risque de fuite de données
variables_leakage = [
    "koi_pdisposition",
    "koi_score",
    "koi_fpflag_nt",
    "koi_fpflag_ss",
    "koi_fpflag_co",
    "koi_fpflag_ec",
    "kepler_name"
]

# 4. Variables d'incertitude
variables_incertitude = [
    col for col in df.columns
    if "_err" in col
]

# 5. Variables catégorielles et descriptives
variables_categorielles = [
    col for col in df.select_dtypes(include=["object"]).columns
    if col != cible
]

# 6. Variables numériques
variables_numeriques = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# 7. Variables de mesure numérique candidates
# Exclusion des identifiants, flags et incertitudes
variables_mesures = [
    col for col in variables_numeriques
    if col not in identifiants
    and col not in variables_leakage
    and col not in variables_incertitude
]

# Affichage des groupes
groupes = {
    "Identifiants": identifiants,
    "Variable cible": [cible],
    "Risque de leakage": variables_leakage,
    "Variables d'incertitude": variables_incertitude,
    "Variables catégorielles": variables_categorielles,
    "Variables numériques": variables_numeriques,
    "Variables de mesure candidates": variables_mesures
}

for nom, colonnes in groupes.items():
    print(f"\n========== {nom} ({len(colonnes)}) ==========")
    print(colonnes)
```

```python

# ============================================================
# STATISTIQUES DESCRIPTIVES
# ============================================================

statistiques = df[variables_numeriques].describe().T

# Ajouter la médiane et le nombre de valeurs manquantes
statistiques["median"] = df[variables_numeriques].median()
statistiques["missing"] = df[variables_numeriques].isna().sum()

# Pourcentage de valeurs manquantes
statistiques["missing_percent"] = (
    statistiques["missing"] / len(df) * 100
).round(2)

# Afficher les résultats
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

display(statistiques.round(3))
```

```python

# ============================================================
# VERIFICATION DES VARIABLES BINAIRES
# ============================================================

variables_flags = [
    "koi_fpflag_nt",
    "koi_fpflag_ss",
    "koi_fpflag_co",
    "koi_fpflag_ec"
]

for col in variables_flags:

    print(f"\n========== {col} ==========")

    print("Valeurs observées :")
    print(df[col].value_counts(dropna=False))

    # Les valeurs autorisées pour un indicateur binaire sont 0 et 1
    valeurs_invalides = df.loc[
        ~df[col].isin([0, 1]),
        [col, cible]
    ]

    print("\nNombre de valeurs hors de {0, 1} :",
          len(valeurs_invalides))

    if not valeurs_invalides.empty:
        display(valeurs_invalides)
```
